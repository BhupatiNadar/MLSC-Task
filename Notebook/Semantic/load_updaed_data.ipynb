```python
import numpy as np

def row_wise_cosine_similarity(A, B):
    A_norm = A / np.linalg.norm(A, axis=1, keepdims=True)
    B_norm = B / np.linalg.norm(B, axis=1, keepdims=True)
    return np.sum(A_norm * B_norm, axis=1)

doc_response_similarity = row_wise_cosine_similarity(document_embeddings, response_embeddings)
question_response_similarity = row_wise_cosine_similarity(question_embeddings, response_embeddings)
question_document_similarity = row_wise_cosine_similarity(question_embeddings, document_embeddings)
```

```python
df["doc_response_similarity"] = doc_response_similarity
df["question_response_similarity"] = question_response_similarity
df["question_document_similarity"] = question_document_similarity
```
